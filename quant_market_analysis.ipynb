```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
aapl=yf.download("AAPL",start="2021-01-01", end="2026-01-01")
nvda=yf.download("NVDA",start="2021-01-01", end="2026-01-01")
aapl.head()
nvda.head()
```

```python
aapl.info()
nvda.info()
```

```python
aapl.describe()
```

```python
aapl_close=aapl["Close"]
```

```python
aapl_close.head()
```

```python
aapl_returns=aapl_close.pct_change()
```

```python
aapl_cumulative_returns=(1+aapl_returns).cumprod()-1
```

```python
aapl_daily_volatility=aapl_returns.std()
```

```python
aapl_daily_volatility
```

```python
aapl_annual_volatility=aapl_daily_volatility * np.sqrt(252)
```

```python
aapl_annual_volatility
```

```python
aapl_sma20=aapl_close.rolling(window=20).mean()
```

```python
aapl_sma20.head(25)
```

```python
aapl_sma50=aapl_close.rolling(window=50).mean()
```

```python
aapl_rolling_volatility=aapl_returns.rolling(window=20).std()
```

```python
aapl_rolling_volatility.tail()
```

```python
aapl_running_max=aapl_close.cummax()
```

```python
aapl_running_max.head(20)
```

```python
aapl_drawdown=(aapl_close/aapl_running_max)-1
```

```python
aapl_drawdown.head(20)
```

```python
aapl_max_drawdown=aapl_drawdown.min()
```

```python
aapl_max_drawdown
```

```python
aapl_drawdown.idxmin()
```

```python
nvda_close=nvda["Close"]
```

```python
nvda_close.head()
```

```python
nvda_returns=nvda_close.pct_change()
```

```python
nvda_returns.head()
```

```python
returns_comparison=pd.concat(
    [aapl_returns, nvda_returns],
    axis=1
)
```

```python
correlation=returns_comparison.corr()
```

```python
correlation
```

```python
aapl_avg_daily_return=aapl_returns.mean()
```

```python
aapl_avg_daily_return
```

```python
aapl_total_return = aapl_cumulative_returns.iloc[-1]
```

```python
aapl_total_return
```

```python
aapl_best_day = aapl_returns.max()
aapl_worst_day = aapl_returns.min()

aapl_best_day, aapl_worst_day
```

```python
aapl_returns.idxmax(), aapl_returns.idxmin()
```

```python
positive_days = (aapl_returns > 0).sum()
negative_days = (aapl_returns < 0).sum()

positive_days, negative_days
```

```python
plt.figure(figsize=(14,6))

plt.plot(aapl_close, label="AAPL Close")
plt.plot(aapl_sma20, label="SMA20")
plt.plot(aapl_sma50, label="SMA50")

plt.title("AAPL Price with SMA20 and SMA50")
plt.xlabel("Date")
plt.ylabel("Price ($)")
plt.legend()
plt.grid()
plt.tight_layout()

plt.savefig("images/aapl_price_sma.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(aapl_cumulative_returns, label="AAPL Cumulative Return")

plt.title("AAPL Cumulative Returns")
plt.xlabel("Date")
plt.ylabel("Cumulative Return")
plt.legend()
plt.grid()
plt.tight_layout()

plt.savefig("images/aapl_cumulative_returns.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(aapl_rolling_volatility, label="20-Day Rolling Volatility")

plt.title("AAPL 20-Day Rolling Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.legend()
plt.grid()
plt.tight_layout()

plt.savefig("images/aapl_rolling_volatility.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(14,6))

plt.plot(aapl_drawdown, label="AAPL Drawdown")

plt.title("AAPL Drawdown")
plt.xlabel("Date")
plt.ylabel("Drawdown")
plt.legend()
plt.grid()
plt.tight_layout()

plt.savefig("images/aapl_drawdown.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
nvda_cumulative_returns = (1 + nvda_returns).cumprod() - 1
```

```python
plt.figure(figsize=(14,6))

plt.plot(aapl_cumulative_returns, label="AAPL")
plt.plot(nvda_cumulative_returns, label="NVDA")

plt.title("AAPL vs NVDA Cumulative Returns")
plt.xlabel("Date")
plt.ylabel("Cumulative Return")
plt.legend()
plt.grid()
plt.tight_layout()

plt.savefig("images/aapl_vs_nvda_returns.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
performance_summary = {
    "Average Daily Return": aapl_avg_daily_return,
    "Total Return": aapl_total_return,
    "Annualized Volatility": aapl_annual_volatility,
    "Maximum Drawdown": aapl_max_drawdown,
    "Best Trading Day": aapl_best_day,
    "Worst Trading Day": aapl_worst_day,
    "Positive Days": positive_days,
    "Negative Days": negative_days,
    "AAPL-NVDA Correlation": correlation.loc["AAPL", "NVDA"]
}

performance_summary
```
